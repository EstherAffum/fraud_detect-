```python
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN

# Initialize Presentation
prs = Presentation()

# Slide 1: Title Slide
slide_layout = prs.slide_layouts[0]
slide = prs.slides.add_slide(slide_layout)
title = slide.shapes.title
subtitle = slide.placeholders[1]

title.text = "Fraud Detection Web App Model"
subtitle.text = "Machine Learning for Real-Time Financial Fraud Detection\nData Science & Engineering Team"

# Helper function to add bullet slides
def add_bullet_slide(prs, slide_title, bullet_points):
    slide_layout = prs.slide_layouts[1]
    slide = prs.slides.add_slide(slide_layout)
    shapes = slide.shapes
    
    title_shape = shapes.title
    title_shape.text = slide_title
    
    tf = shapes.placeholders[1].text_frame
    tf.word_wrap = True
    
    for idx, pt in enumerate(bullet_points):
        if idx == 0:
            p = tf.paragraphs[0]
        else:
            p = tf.add_paragraph()
        
        if isinstance(pt, tuple):
            p.text = pt[0]
            p.level = pt[1]
        else:
            p.text = pt
            p.level = 0

# Slide Data
slides_data = [
    ("Executive Summary & Objectives", [
        ("Core Problem: Financial fraud creates significant financial loss and undermines trust.", 0),
        ("Objective: Build a real-time web application flagging suspicious transactions.", 0),
        ("Key Deliverables:", 0),
        ("Trained machine learning classification pipeline", 1),
        ("Interactive web dashboard for inference and analytics", 1),
        ("REST API endpoint for payment processing integration", 1),
    ]),
    ("Data Pipeline & Preprocessing", [
        ("Dataset Characteristics: Highly imbalanced transaction data (< 1% fraud cases).", 0),
        ("Preprocessing Steps:", 0),
        ("Feature Scaling: Standardizing transaction amounts and temporal features", 1),
        ("Class Imbalance: Applied SMOTE / Random Under-sampling", 1),
        ("Feature Selection: Correlation analysis and feature importance filtering", 1),
    ]),
    ("Model Architecture & Evaluation", [
        ("Algorithms Evaluated: Logistic Regression, Random Forest, XGBoost", 0),
        ("Key Metrics:", 0),
        ("Recall: Prioritized to minimize missed fraudulent activities", 1),
        ("Precision & F1-Score: Balanced to avoid false customer flags", 1),
        ("ROC-AUC & PR-AUC: Evaluated across variable probability thresholds", 1),
    ]),
    ("Web Application Architecture", [
        ("Frontend: Streamlit / React interface for analytical monitoring", 0),
        ("Backend API: FastAPI / Flask serving low-latency predictions", 0),
        ("Model Persistence: Serialized model artifacts (.joblib / .pkl)", 0),
        ("Deployment: Containerized via Docker for cloud scalability", 0),
    ]),
    ("Results & Deployment Roadmap", [
        ("Performance: Sub-100ms inference time with high fraud detection recall.", 0),
        ("Business Impact: Reduction in fraudulent transaction pass-through.", 0),
        ("Future Enhancements:", 0),
        ("Streaming data integration via Apache Kafka", 1),
        ("Automated continuous retraining pipelines (MLOps)", 1),
        ("Unsupervised anomaly detection with Deep Autoencoders", 1),
    ])
]

# Add content slides
for title, points in slides_data:
    add_bullet_slide(prs, title, points)

# Save presentation
prs.save("fraud_detection_presentation.pptx")
print("Presentation saved successfully as 'fraud_detection_presentation.pptx'.")
```
